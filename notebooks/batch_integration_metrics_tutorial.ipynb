# Batch Integration Metrics Tutorial

Quick synthetic example for the metric functions in `ne_aligned_utils.metrics`.

```python
import numpy as np
import pandas as pd

from ne_aligned_utils.metrics import (
    compute_ilisi,
    compute_kbet,
    compute_kmeans_cluster_scores,
    compute_knn_other_batch_fraction,
    compute_neighbor_purity,
    compute_silhouette_index,
    compute_silhouette_score,
)

rng = np.random.default_rng(0)
n_per_label_batch = 60
centers = np.array([[-2, 0], [2, 0]], dtype=float)
embedding_parts = []
labels = []
batch_ids = []
for label, center in enumerate(centers):
    for batch, offset in enumerate([[-0.1, 0], [0.1, 0]]):
        embedding_parts.append(rng.normal(center + offset, 0.45, size=(n_per_label_batch, 2)))
        labels.extend([label] * n_per_label_batch)
        batch_ids.extend([batch] * n_per_label_batch)
embedding = np.vstack(embedding_parts)
labels = np.asarray(labels)
batch_ids = np.asarray(batch_ids)
```

```python
cluster_scores = compute_kmeans_cluster_scores(embedding, labels)
pd.DataFrame([
    {'metric': 'neighbor_purity', 'value': compute_neighbor_purity(embedding, labels)},
    {'metric': 'silhouette_index', 'value': compute_silhouette_index(embedding, labels)},
    {'metric': 'scaled_silhouette_score', 'value': compute_silhouette_score(embedding, labels)},
    {'metric': 'iLISI', 'value': compute_ilisi(embedding, batch_ids)},
    {'metric': 'kBET', 'value': compute_kbet(embedding, batch_ids, labels)},
    {'metric': 'shared_cell_type_batch_mixing', 'value': compute_knn_other_batch_fraction(embedding, batch_ids, labels)},
    {'metric': 'kmeans_rand_index', 'value': cluster_scores['rand_index']},
    {'metric': 'kmeans_adjusted_rand_index', 'value': cluster_scores['adjusted_rand_index']},
]).round(4)
```
